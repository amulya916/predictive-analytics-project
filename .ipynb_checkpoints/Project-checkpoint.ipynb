```python
numeric_cols = df.select_dtypes(include=['int64','float64']).columns
cat_cols = df.select_dtypes(include='object')
```

Check for outliers

```python
import math
cols = 3
rows =math.ceil(len(numeric_cols) / cols)
plt.figure(figsize=(15, rows * 4))
for i, col in enumerate(numeric_cols):
    plt.subplot(rows,cols, i + 1)
    sns.boxplot(y=df[col])
    plt.title(f'Boxplot of {col}')
    plt.ylabel('') 
plt.show()
```

```python
def remove_outliers_iqr(df, cols):
    for col in cols:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        df = df[(df[col] >= lower) & (df[col] <= upper)]     
    return df

df = remove_outliers_iqr(df, numeric_cols)
df.shape
```

```python
df['Yield_Category'] = pd.qcut(
    df['Paddy yield(in Kg)'],
    q=3,
    labels=['Low', 'Medium', 'High']
)
```

```python
x_yield = df.drop(['Paddy yield(in Kg)', 'Yield_Category'], axis=1)
y_yield = df['Yield_Category']

x_train, x_test, y_train, y_test = train_test_split(x_yield, y_yield, test_size=0.2, random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

rfc = RandomForestClassifier(n_estimators=200, random_state=42)
rfc.fit(x_train, y_train)

rfc_pred = rfc.predict(x_test)
```

```python
print("Accuracy:", accuracy_score(y_test, rfc_pred))
print("Confusion Matrix:", confusion_matrix(y_test, rfc_pred))
print(classification_report(y_test, rfc_pred))
```

```python
importance = pd.Series(
    rf.feature_importances_,
    index=xtrain.columns
).sort_values(ascending=False)

importance.head(5)
```

```python
plt.figure(figsize=(4,3))
importance.head(5).plot(kind='bar')
plt.title("Top Factors Affecting Paddy Yield")
plt.show()
```

```python
from scipy.stats import zscore

df['yield_zscore'] = zscore(df['Paddy yield(in Kg)'])

anomalies = df[abs(df['yield_zscore']) > 3]
print("Anomalies detected:", anomalies.shape[0])
```

```python
plt.figure()
plt.scatter(range(len(df)), df['Paddy yield(in Kg)'], label='Normal')
plt.scatter(anomalies.index, anomalies['Paddy yield(in Kg)'], color='red', label='Anomaly')
plt.legend()
plt.show()
```
